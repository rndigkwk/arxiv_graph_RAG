# data 폴더 jsonl 파일 컬럼 정리

`data/` 폴더의 모든 `.jsonl` 파일에서 `updated`, `doi`, `pdf_url`, `source`, `collection_window` 컬럼을 제거하고,
다른 이름(`*_cleaned.jsonl`)으로 `data/` 폴더에 저장합니다.

```python
import glob
import json
import os

import pandas as pd

DATA_DIR = "../data"
COLUMNS_TO_DROP = ["updated", "doi", "pdf_url", "source", "collection_window"]
OUTPUT_SUFFIX = "_cleaned"
```

```python
jsonl_paths = sorted(glob.glob(os.path.join(DATA_DIR, "*.jsonl")))
# 이미 정리된 파일은 다시 처리하지 않도록 제외
jsonl_paths = [p for p in jsonl_paths if not os.path.basename(p).endswith(f"{OUTPUT_SUFFIX}.jsonl")]
jsonl_paths
```

```python
for path in jsonl_paths:
    df = pd.read_json(path, lines=True)

    existing_cols = [c for c in COLUMNS_TO_DROP if c in df.columns]
    df = df.drop(columns=existing_cols)

    base, ext = os.path.splitext(path)
    output_path = f"{base}{OUTPUT_SUFFIX}{ext}"

    # pandas.to_json escapes "/" as "\/"; use the json module instead to keep it intact.
    with open(output_path, "w", encoding="utf-8") as f:
        for record in df.to_dict(orient="records"):
            f.write(json.dumps(record, ensure_ascii=False))
            f.write("\n")

    print(f"{path} -> {output_path} ({len(df)} rows, dropped {existing_cols})")
```
